## 2. The question: decision, action, cost of a wrong call

## 2. The question: decision, action, cost of a wrong call

* **Research Question:** Which existing content items with sufficient search visibility show observable signs of traffic decline or decay risk and should be prioritized first for editorial refresh review?
* **Unit of Analysis (Grain):** One individual content item (`content_id` / `content_hash_id`) per evaluation window.
* **Decision & Action:** Produce a ranked review queue (e.g., Top-50 candidates) with human-readable reason codes so an SEO or content editor can decide whether to update, expand, consolidate, or monitor each flagged page.
* **Cost of Getting It Wrong:**
  * *False Positive (flagging a healthy page):* Wastes limited writer and editor hours updating content that did not need intervention, and risks unintentionally hurting stable rankings.
  * *False Negative (missing a truly declining high-demand page):* Continued loss of organic impressions and clicks on valuable pages that could have been stabilized earlier.
* **Why Data/ML Helps Beyond Simple Rules:** Simple heuristics (such as flagging any old page or any page with a single-metric dip) produce high false-positive rates (`0.240` Precision@50 in the starter baseline). A supervised ranking model can weigh multi-signal interactions—visibility tier, position, CTR, engagement rate, and content age—simultaneously.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os, subprocess, pandas as pd

# Ensure the starter repo/CSV is present in the Colab runtime
if not os.path.exists("data/raw/content_refresh_anonymized.csv"):
    if not os.path.isdir("flyrank-ml-internship-starter"):
        subprocess.run(["git", "clone", "--depth", "1", "https://github.com/flyrank-bih/flyrank-ml-internship-starter"], check=True)
    os.chdir("flyrank-ml-internship-starter")

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

total_rows = len(df)
eligible = df[(df["impressions_90d"] > 0) & (df["content_age_days"] >= 90)]
declining_rate = (eligible["trend_direction"] == "down").mean()
median_impr_down = eligible[eligible["trend_direction"] == "down"]["impressions_90d"].median()
median_impr_up = eligible[eligible["trend_direction"] == "up"]["impressions_90d"].median()

print(f"1. Total rows in starter dataset: {total_rows:,}")
print(f"2. Eligible mature pages (impressions > 0, age >= 90d): {len(eligible):,} ({len(eligible)/total_rows:.1%})")
print(f"3. Observed declining ('down') share among eligible pages: {declining_rate:.1%}")
print(f"4. Median 90d impressions — declining pages: {median_impr_down:,.0f} vs. growing ('up') pages: {median_impr_up:,.0f}")
```

## 4. Careful words: what I can and can't claim

* **Observed & Directional Only:** All findings from the starter slice and warehouse tables are observational. A high priority score or observed correlation indicates that a page resembles historically declining pages—not that an algorithm rule has been proven.
* **No Causal Guarantee:** Ranking a page as a top refresh candidate identifies where human review is best spent; without an A/B experiment or controlled intervention log, this data cannot prove that editing a page *caused* a subsequent traffic recovery.
* **Scope & Leakage Discipline:** The starter dataset uses a current-window proxy label (`trend_direction == "down"`). When scaling to the full Hugging Face warehouse (`FlyRank/internship-warehouse`), I will separate prior feature windows from future target windows (`prior 90d -> next 30d`), enforce client-grouped splits, and rule out consolidation or seasonal dips.
